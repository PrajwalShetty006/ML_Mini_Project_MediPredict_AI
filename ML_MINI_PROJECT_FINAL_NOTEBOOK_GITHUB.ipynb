```python
print("DISEASE DISTRIBUTION")

disease_counts = df['Disease'].value_counts()
print(disease_counts.head(10))

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

disease_counts.head(25).plot(kind='barh', ax=axes[0])
axes[0].set_title('Top 25 Diseases')
axes[0].invert_yaxis()

axes[1].hist(disease_counts.values, bins=30)
axes[1].set_title('Disease Frequency Distribution')

plt.tight_layout()
plt.show()
```

```python

symptoms = set()

for col in df.columns[1:]:
    symptoms.update(df[col].unique())

symptoms.discard(0)

new_df = pd.DataFrame(columns=list(symptoms))
new_df["Disease"] = df["Disease"]

new_df.fillna(0,inplace=True)

for i in range(len(df)):
    for col in df.columns[1:]:
        symptom = df.iloc[i][col]
        if symptom != 0:
            new_df.at[i, symptom] = 1

new_df.fillna(0,inplace=True)

print(new_df.head())
```

```python
new_df
```

```python
cols = list(new_df.columns)
cols.remove("Disease")
new_df = new_df[["Disease"] + cols]

new_df
```

```python
new_df
```

```python
symptom_features = new_df.columns[1:]
symptom_frequencies = new_df[symptom_features].sum().sort_values(ascending=False)
print(f"Total Symptoms: {len(symptom_frequencies)}")
print(f"\nTop 20 Most Common Symptoms:")
print(symptom_frequencies.head(20))
print(f"\nRarest 10 Symptoms:")
print(symptom_frequencies.tail(10))
fig, axes = plt.subplots(2, 2, figsize=(18, 12))

symptom_frequencies.head(30).plot(kind='barh', ax=axes[0, 0], color='skyblue')
axes[0, 0].set_title('Top 30 Most Frequent Symptoms', fontsize=14, fontweight='bold')
axes[0, 0].set_xlabel('Frequency')
axes[0, 0].set_ylabel('Symptom')
axes[0, 0].invert_yaxis()

symptom_frequencies.tail(30).plot(kind='barh', ax=axes[0, 1], color='salmon')
axes[0, 1].set_title('30 Least Frequent Symptoms', fontsize=14, fontweight='bold')
axes[0, 1].set_xlabel('Frequency')
axes[0, 1].set_ylabel('Symptom')
axes[0, 1].invert_yaxis()

axes[1, 0].hist(symptom_frequencies.values, bins=50, color='purple',
                edgecolor='black', alpha=0.7)
axes[1, 0].set_title('Distribution of Symptom Frequencies', fontsize=14, fontweight='bold')
axes[1, 0].set_xlabel('Frequency')
axes[1, 0].set_ylabel('Number of Symptoms')
axes[1, 0].grid(True, alpha=0.3)
axes[1, 0].axvline(symptom_frequencies.mean(), color='red',
                   linestyle='--', linewidth=2, label=f'Mean: {symptom_frequencies.mean():.1f}')
axes[1, 0].axvline(symptom_frequencies.median(), color='green',
                   linestyle='--', linewidth=2, label=f'Median: {symptom_frequencies.median():.1f}')
axes[1, 0].legend()

stats_df = pd.DataFrame({
    'Statistic': ['Mean', 'Median', 'Std Dev', 'Min', 'Max', 'Q1', 'Q3'],
    'Value': [symptom_frequencies.mean(), symptom_frequencies.median(),
              symptom_frequencies.std(), symptom_frequencies.min(),
              symptom_frequencies.max(), symptom_frequencies.quantile(0.25),
              symptom_frequencies.quantile(0.75)]
})
axes[1, 1].axis('tight')
axes[1, 1].axis('off')
table = axes[1, 1].table(cellText=stats_df.values, colLabels=stats_df.columns,
                         cellLoc='center', loc='center', colWidths=[0.5, 0.3])
table.auto_set_font_size(False)
table.set_fontsize(12)
table.scale(1, 2.5)
axes[1, 1].set_title('Symptom Frequency Statistics', fontsize=14, fontweight='bold', pad=20)

plt.tight_layout()
plt.show()
```

```python
print(" SYMPTOM CORRELATION ANALYSIS")

top_symptoms = symptom_frequencies.head(20).index
correlation_matrix = new_df[top_symptoms].corr()
plt.figure(figsize=(14, 12))
sns.heatmap(correlation_matrix, annot=True, fmt='.2f', cmap='coolwarm',
            center=0, square=True, linewidths=0.5, cbar_kws={"shrink": 0.8})
plt.title('Correlation Heatmap: Top 20 Most Frequent Symptoms',
          fontsize=16, fontweight='bold', pad=20)
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()

high_corr_pairs = []
for i in range(len(correlation_matrix.columns)):
    for j in range(i+1, len(correlation_matrix.columns)):
        if abs(correlation_matrix.iloc[i, j]) > 0.5:
            high_corr_pairs.append({
                'Symptom 1': correlation_matrix.columns[i],
                'Symptom 2': correlation_matrix.columns[j],
                'Correlation': correlation_matrix.iloc[i, j]
            })

if high_corr_pairs:
    high_corr_df = pd.DataFrame(high_corr_pairs).sort_values('Correlation', ascending=False)
    print(f"\nHighly Correlated Symptom Pairs (|r| > 0.5):")
    print(high_corr_df)
else:
    print("\nNo highly correlated symptom pairs found (|r| > 0.5)")
```

```python
X = new_df.drop("Disease", axis=1)
y = new_df["Disease"]
```

```python
print("DATA SPARSITY ANALYSIS")
total_elements = X.shape[0] * X.shape[1]
zero_elements = (X == 0).sum().sum()
sparsity = (zero_elements / total_elements) * 100
print(f"Total Elements: {total_elements:,}")
print(f"Zero Elements: {zero_elements:,}")
print(f"Non-Zero Elements: {total_elements - zero_elements:,}")
print(f"Sparsity: {sparsity:.2f}%")

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

sizes = [zero_elements, total_elements - zero_elements]
labels = ['Zero (Absent)', 'One (Present)']
colors = ['lightcoral', 'lightgreen']
explode = (0.1, 0)
axes[0].pie(sizes, explode=explode, labels=labels, colors=colors, autopct='%1.1f%%',
            startangle=90, shadow=True)
axes[0].set_title('Feature Matrix Sparsity', fontsize=14, fontweight='bold')

axes[1].bar(labels, sizes, color=colors, edgecolor='black')
axes[1].set_title('Zero vs Non-Zero Elements', fontsize=14, fontweight='bold')
axes[1].set_ylabel('Count')
axes[1].ticklabel_format(style='plain', axis='y')
for i, v in enumerate(sizes):
    axes[1].text(i, v, f'{v:,}\n({v/total_elements*100:.1f}%)',
                 ha='center', va='bottom', fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
print(" DISEASE-WISE SYMPTOM PATTERN ANALYSIS")
print("-"*80)

disease_symptom_avg = pd.DataFrame({
    'Disease': new_df['Disease'],
    'Symptom_Count': new_df[X.columns].sum(axis=1)
}).groupby('Disease')['Symptom_Count'].mean().sort_values(ascending=False)

print(f"Diseases with Most Symptoms (Average):")
print(disease_symptom_avg.head(10))
print(f"\nDiseases with Fewest Symptoms (Average):")
print(disease_symptom_avg.tail(10))

fig, axes = plt.subplots(1, 2, figsize=(18, 6))

disease_symptom_avg.head(20).plot(kind='barh', ax=axes[0], color='steelblue')
axes[0].set_title('Top 20 Diseases by Average Symptom Count', fontsize=14, fontweight='bold')
axes[0].set_xlabel('Average Number of Symptoms')
axes[0].set_ylabel('Disease')
axes[0].invert_yaxis()
axes[0].grid(True, alpha=0.3, axis='x')

axes[1].hist(disease_symptom_avg.values, bins=30, color='orange',
             edgecolor='black', alpha=0.7)
axes[1].set_title('Distribution of Average Symptoms per Disease', fontsize=14, fontweight='bold')
axes[1].set_xlabel('Average Number of Symptoms')
axes[1].set_ylabel('Number of Diseases')
axes[1].axvline(disease_symptom_avg.mean(), color='red',
                linestyle='--', linewidth=2, label=f'Mean: {disease_symptom_avg.mean():.2f}')
axes[1].axvline(disease_symptom_avg.median(), color='green',
                linestyle='--', linewidth=2, label=f'Median: {disease_symptom_avg.median():.2f}')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)
```

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(
    n_estimators=100,
    max_depth=10,
    min_samples_split=5,
    min_samples_leaf=3,
    random_state=32
)
model.fit(X_train, y_train)
```

```python
y_pred=model.predict(X_test)
```

```python
rf_acc = model.score(X_test, y_test)
print("Accuracy:", rf_acc)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)
```

```python
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

svc_model = SVC(
    kernel='rbf',
    C=0.5,
    gamma=0.01,
    decision_function_shape='ovr',
    random_state=35
)

svc_model.fit(X_train_scaled, y_train)

svc_pred = svc_model.predict(X_test_scaled)

svc_acc = accuracy_score(y_test, svc_pred)
print("SVM Accuracy:", svc_acc)
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

dt_model = DecisionTreeClassifier(
    max_depth=50,
    min_samples_split=20,
    min_samples_leaf=10,
    random_state=42
)

dt_model.fit(X_train, y_train)

dt_pred = dt_model.predict(X_test)
dt_acc = accuracy_score(y_test, dt_pred)

print("Decision Tree Accuracy:", dt_acc)
```

```python
import matplotlib.pyplot as plt

models = ['Decision Tree', 'Random Forest', 'SVM']
accuracies = [dt_acc, rf_acc, svc_acc]

plt.figure()
plt.plot(models, accuracies, marker='o',)

plt.xlabel("Models")
plt.ylabel("Accuracy")
plt.title("Accuracy Comparison of Models")

for i, v in enumerate(accuracies):
    plt.text(i, v, str(round(v, 3)))

plt.grid()

plt.show()
```
